## 📖 Descripción de la Asignatura

El **Aprendizaje Automático Avanzado** es una disciplina que parte de los **algoritmos avanzados del Machine Learning clásico** y avanza hacia el **Deep Learning** moderno.
> Comenzamos con un repaso del aprendizaje **supervisado** y **no supervisado**, y con **algoritmos avanzados** para datos tabulares y de agrupamiento: **Gradient Boosting** (XGBoost, LightGBM, CatBoost), **clustering basado en densidad** (DBSCAN, HDBSCAN, OPTICS) y **clustering probabilístico** (Gaussian Mixture Models y mezclas bayesianas). A partir de ahí, la idea central es dotar a las máquinas de la capacidad de **aprender representaciones** directamente de los datos mediante **redes neuronales profundas**. A lo largo del curso construiremos, entrenaremos y evaluaremos arquitecturas usando **TensorFlow 2.0/Keras** y **PyTorch**: redes densas (MLP), **redes convolucionales (CNN)** para visión, **redes recurrentes** (RNN, LSTM, GRU) y modelos basados en **atención y Transformers** para secuencias, **modelos generativos** (Autoencoders y VAE) y agentes de **aprendizaje por refuerzo profundo**.

## 📓 ¿Qué es un Jupyter Notebook?

Un Jupyter Notebook es un documento interactivo que combina código ejecutable, texto enriquecido (Markdown), ecuaciones matemáticas y visualizaciones. Es ideal para prototipar, analizar datos y compartir resultados.

**Tipos de celdas:**
- **Código:** Ejecuta instrucciones en Python y muestra la salida.
- **Markdown:** Permite escribir texto, ecuaciones y agregar imágenes.

**Ejemplo de celda de código:**

````python
# Ejemplo: suma de dos números
a = 5
b = 3
print(f"La suma de {a} + {b} es {a + b}")
````

**Ejemplo de ecuación en Markdown (salida de una neurona):**

$$\hat{y} = \sigma\!\left(\sum_{i=1}^{n} w_i x_i + b\right)$$

**Descenso de gradiente (actualización de parámetros):**

$$\theta \leftarrow \theta - \eta \, \nabla_{\theta} \, \mathcal{L}(\theta)$$

donde $\eta$ es la tasa de aprendizaje y $\mathcal{L}(\theta)$ la función de pérdida.

## 6. Configuración del Entorno

Para trabajar en Deep Learning con Python usaremos un **entorno de conda aislado** llamado `aaa_curso_2026`. La forma **recomendada** es crearlo desde el archivo `environment.yml` (Opción A), que ya deja todo listo y reproducible.

> 💻 **Instalación en CPU:** en este curso PyTorch y TensorFlow se instalan en su versión **CPU** (suficiente para la Introducción y la mayoría de las prácticas). En Windows, las ruedas de PyTorch de PyPI ya son CPU por defecto, por eso se instalan vía `pip`.

### Opción A: Crear el entorno desde `environment.yml` (Recomendado)

Crea (o usa el que ya viene en el curso) un archivo `environment.yml` con este contenido:

```yaml
name: aaa_curso_2026

channels:
  - conda-forge
  - defaults

dependencies:
  # Python
  - python=3.10.12

  # Jupyter
  - jupyterlab>=4.0
  - ipykernel>=6.29
  - ipywidgets>=8.1

  # Machine Learning clásico
  - scikit-learn>=1.4
  - imbalanced-learn>=0.11

  # Deep Learning: PyTorch (CPU) se instala vía pip más abajo

  # Data
  - numpy>=1.26   # TF y PyTorch (CPU) recientes son compatibles con NumPy 2.x
  - pandas>=2.1
  - scipy>=1.11

  # Visualización
  - matplotlib>=3.8
  - seaborn>=0.13
  - plotly>=5.18

  # Utilidades
  - tqdm>=4.66
  - requests>=2.31
  - python-dotenv>=1.0

  # Paquetes vía pip (no disponibles o más actualizados en pip)
  - pip>=24.0
  - pip:
    # PyTorch build CPU (en Windows, las ruedas de PyPI ya son CPU)
    - torch>=2.1
    - torchvision>=0.16
    - torchaudio>=2.1
    - tensorflow>=2.15       # framework de Deep Learning (Keras incluido)
    - tensorboard>=2.15      # seguimiento de experimentos
    - optuna>=3.5            # búsqueda de hiperparámetros
    - torchinfo>=1.8         # resumen de modelos
    - gymnasium>=0.29        # entornos de aprendizaje por refuerzo
    - transformers>=4.38     # arquitecturas Transformer (Módulo 3)
    - datasets>=2.17
    - grad-cam>=1.5          # interpretabilidad de CNN (Módulo 2)
    - opencv-python>=4.9
    - pillow>=10.2
```

> ⚠️ **Importante:** los paquetes de `pip` deben ir **indentados** bajo la clave `- pip:`. Si los dejas sueltos, el archivo YAML será inválido.

### Opción B: Instalación manual con conda + pip

Si prefieres construir el entorno paso a paso (sin el archivo `environment.yml`):

```bash
# Crear y activar el entorno
conda create -n aaa_curso_2026 python=3.10
conda activate aaa_curso_2026

# Librerías base desde conda-forge
conda install -c conda-forge jupyterlab ipykernel ipywidgets numpy pandas scipy matplotlib seaborn plotly scikit-learn imbalanced-learn tqdm requests python-dotenv

# PyTorch (CPU) y TensorFlow vía pip
pip install torch torchvision torchaudio
pip install tensorflow

# Herramientas de Deep Learning y experimentación
pip install tensorboard optuna torchinfo gymnasium transformers datasets grad-cam opencv-python pillow
```

> 🐧 En **Linux/Mac**, para forzar PyTorch en CPU usa: `pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu`


### Procedimiento para instalar el entorno del curso

Estos son los pasos exactos que seguimos para crear y verificar el entorno:

```bash
# 1. Abre Anaconda Prompt (Windows) o Terminal (Mac/Linux)
# 2. Navega a la carpeta del curso donde está environment.yml:
cd "ruta/al/curso"

# 3. Crea el entorno desde el archivo (instala conda + pip automáticamente)
conda env create -f environment.yml

# 4. Activa el entorno
conda activate aaa_curso_2026

# 5. Verifica PyTorch (debe reportar versión '+cpu' y GPU disponible: False)
python -c "import torch; print('PyTorch', torch.__version__, '| CUDA disponible:', torch.cuda.is_available())"

# 6. Verifica TensorFlow
python -c "import tensorflow as tf; print('TensorFlow', tf.__version__, '| GPUs:', len(tf.config.list_physical_devices('GPU')))"

# 7. Registra el kernel en Jupyter para poder seleccionarlo en VS Code
python -m ipykernel install --user --name aaa_curso_2026 --display-name "Python (AAA 2026)"

# 8. Inicia Jupyter (o abre los notebooks directamente en VS Code)
jupyter notebook
```

**Salida esperada de la verificación** (las versiones pueden variar ligeramente):

```
PyTorch 2.13.0+cpu | CUDA disponible: False
TensorFlow 2.21.0 | GPUs: 0
```

> ✅ El sufijo `+cpu` y `CUDA disponible: False` confirman que PyTorch está instalado en su **versión CPU**, como buscamos en este curso.

